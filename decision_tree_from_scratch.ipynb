```python
def gini_index(count_of_class):
    # Calculate the total number of instances
    total_instances = sum(count_of_class.values())
    # Calculate the sum of squared proportions for each class
    sum_of_squared_proportions = sum((n / total_instances) ** 2 for n in count_of_class.values())
    # Calculate the Gini index
    gini_value = 1 - sum_of_squared_proportions
    return gini_value
```

```python
def gini_split(count_of_class_parent, count_of_class_left, count_of_class_right):
    # Calculate total instances in parent, left, and right
    total_parent = sum(count_of_class_parent.values())
    total_left = sum(count_of_class_left.values())
    total_right = sum(count_of_class_right.values())

    # Calculate Gini index for left and right children
    gini_left = gini_index(count_of_class_left)
    gini_right = gini_index(count_of_class_right)

    # Calculate weighted Gini split
    gini_split_value = (total_left / total_parent) * gini_left + (total_right / total_parent) * gini_right

    return gini_split_value
```

```python
def data_split(data, threshold, dim):
    data1 = {}
    data2 = {}

    # Iterate over each class in the dataset
    for class_label, points in data.items():
        # Split points based on the threshold and dimension
        left_points = [point for point in points if point[dim] <= threshold]
        right_points = [point for point in points if point[dim] > threshold]

        # Add the split points to the corresponding new datasets
        data1[class_label] = left_points
        data2[class_label] = right_points

    return data1, data2
```
